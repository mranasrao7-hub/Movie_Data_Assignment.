### **Question 24**

### **Question 26**
(Visualization) Dataset mein 'Certificate' wise average 'Gross' kamai ka ek Bar Plot banayein.

```python
dta['Gross'] = pd.to_numeric(
    dta['Gross'].astype(str).str.replace(',', ''),
    errors='coerce'
)

avg_gross = dta.groupby('Certificate')['Gross'].mean().dropna()

# Bar Plot
avg_gross.plot(kind='bar', figsize=(10, 5))

plt.title('Average Gross Revenue by Certificate')
plt.xlabel('Certificate')
plt.ylabel('Average Gross Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### **Question 27**
(Visualization) Har saal (Released_Year) kitni movies release hueen, iska ek Line Plot banayein.

```python
# Released_Year ko numeric banana
year = pd.to_numeric(dta['Released_Year'], errors='coerce')

# Har saal movies ka count
movies_per_year = year.value_counts().sort_index()
# Line Plot
plt.figure(figsize=(12, 6))

plt.plot(movies_per_year.index, movies_per_year.values, marker='o')

plt.title('Number of Movies Released Each Year')
plt.xlabel('Released Year')
plt.ylabel('Number of Movies')
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

### **Question 36**
(Visualization) 'Drama' aur 'Action' genres ki movies ke 'Gross' revenue ko compare karne ke liye ek Grouped Box Plot ya Violin Plot banayein.

```python
# Gross ko numeric banana
dta['Gross'] = pd.to_numeric(
    dta['Gross'].astype(str).str.replace(',', ''),
    errors='coerce'
)

# Drama aur Action movies filter karna
drama = dta[dta['Genre'].str.contains('Drama', na=False)]['Gross'].dropna()
action = dta[dta['Genre'].str.contains('Action', na=False)]['Gross'].dropna()

# Grouped Box Plot
plt.figure(figsize=(9, 5))

plt.boxplot(
    [drama, action],
    tick_labels=['Drama', 'Action']
)

plt.title('Gross Revenue: Drama vs Action Movies')
plt.xlabel('Genre')
plt.ylabel('Gross Revenue')

plt.tight_layout()
plt.show()
```

### **Question 40**
(Visualization) Top 5 highest rated movies aur unke 'Meta_score' vs 'IMDB_Rating' (scaled to 100) ka ek Side-by-Side Comparison Bar Chart banayein.

```python
import numpy as np

# Top 5 highest-rated movies
top5 = dta.nlargest(5, 'IMDB_Rating').copy()

# IMDB rating ko 100 ke scale par convert karna
top5['IMDB_Scaled'] = top5['IMDB_Rating'] * 10

# Bar chart ki positions
x = np.arange(len(top5))
width = 0.35

plt.figure(figsize=(12, 6))

plt.bar(x - width/2, top5['Meta_score'], width, label='Meta Score')
plt.bar(x + width/2, top5['IMDB_Scaled'], width, label='IMDB Rating (Scaled to 100)')

plt.xticks(x, top5['Series_Title'], rotation=45, ha='right')
plt.ylabel('Score (out of 100)')
plt.title('Top 5 Movies: Meta Score vs IMDB Rating')
plt.legend()
plt.tight_layout()
plt.show()
```
